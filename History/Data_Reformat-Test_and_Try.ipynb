```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

cv = TimeSeriesSplit(n_splits=5)

param_grid = {
    "max_iter": [100, 500, 1000, 10_000, 100_000],
    "learning_rate": [0.00001, 0.0001, 0.001, 0.01, 0.1, 1],
    "max_depth": [1, 2, 3, 5, 7],
    "min_samples_leaf": [5, 10, 20, 30],
    "l2_regularization": [0.0, 0.1, 1.0, 3.0, 5.0],
    "max_leaf_nodes": [8, 15, 31]
}

gs = GridSearchCV(
    estimator=HistGradientBoostingRegressor(random_state=42, early_stopping=True, validation_fraction=0.15, n_iter_no_change=15),
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    n_jobs=-1,
    verbose=1,
)
gs.fit(X_train, y_train)

best_model = gs.best_estimator_
y_pred = best_model.predict(X_test)

print(f"Best params: {gs.best_params_}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
print(f"MAE:  {mean_absolute_error(y_test, y_pred):.4f}")
print(f"R²:   {r2_score(y_test, y_pred):.4f}")
```

```python
"""def backward_elimination(X, y, threshold=0.05):
    cols = list(X.columns)
    while True:
        X_sm = sm.add_constant(X[cols])
        model = sm.OLS(y, X_sm).fit()
        max_pval = model.pvalues.drop("const").max()
        if max_pval > threshold:
            worst = model.pvalues.drop("const").idxmax()
            cols.remove(worst)
            print(f"Dropped {worst} (p={max_pval:.4f})")
        else:
            break
    return cols, model"""

import statsmodels.api as sm

def backward_elimination(X, y, verbose=True):
    cols = list(X.columns)
    X_sm = sm.add_constant(X)
    prev_model = sm.OLS(y, X_sm).fit()
    prev_r2 = prev_model.rsquared_adj

    if verbose:
        print(f"Starting Adj R²: {prev_r2:.6f}")
        print("-" * 60)

    while len(cols) > 1:
        X_sm = sm.add_constant(X[cols])
        model = sm.OLS(y, X_sm).fit()
        pvals = model.pvalues.drop("const")
        worst_feat = pvals.idxmax()
        worst_pval = pvals.max()

        trial_cols = [c for c in cols if c != worst_feat]
        X_trial = sm.add_constant(X[trial_cols])
        trial_model = sm.OLS(y, X_trial).fit()
        trial_r2 = trial_model.rsquared_adj

        if trial_r2 < prev_r2:
            if verbose:
                print(f"STOP — dropping {worst_feat} (p={worst_pval:.4f}) would lower Adj R² from {prev_r2:.6f} to {trial_r2:.6f}")
            break

        if verbose:
            print(f"Dropped {worst_feat:20s} | p={worst_pval:.4f} | Adj R²: {prev_r2:.6f} → {trial_r2:.6f}")

        cols = trial_cols
        prev_r2 = trial_r2
        prev_model = trial_model

    if verbose:
        print("-" * 60)
        print(f"Final features: {cols}")
        print(prev_model.summary())

    return cols, prev_model

X = new_df.drop(columns=['DATE', 'DXY_Diff', 'DXY', 'DXY_next'])
y = new_df['DXY_Diff']

split = int(len(df) * 0.75)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

kept_features, final_model = backward_elimination(X_train, y_train)
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
import numpy as np
import itertools

best_aic = np.inf
best_order = None
best_sarimax = None

for p, d, q in itertools.product(range(4), range(2), range(4)):
    try:
        mod = SARIMAX(y_train.values, exog=X_train, order=(p, d, q),
                      enforce_stationarity=False, enforce_invertibility=False)
        res = mod.fit(disp=False, maxiter=200)
        if res.aic < best_aic:
            best_aic = res.aic
            best_order = (p, d, q)
            best_sarimax = res
    except Exception:
        continue

y_pred_sarimax = best_sarimax.forecast(steps=len(y_test), exog=X_test)

print("SARIMAX")
print(f"Best order (p,d,q): {best_order}")
print(f"AIC:  {best_aic:.2f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_sarimax)):.4f}")
print(f"MAE:  {mean_absolute_error(y_test, y_pred_sarimax):.4f}")
print(f"R²:   {r2_score(y_test, y_pred_sarimax):.4f}")

result_df = new_df.iloc[split:].copy()

result_df["predicted"] = np.exp(y_pred)*result_df["DXY"]
result_df

from sklearn.metrics import mean_squared_error

rmse_naive = np.sqrt(mean_squared_error(result_df["DXY_next"], result_df["DXY"]))
rmse_model = np.sqrt(mean_squared_error(result_df["DXY_next"], result_df["predicted"]))

print(f"Naive (DXY_next = DXY) RMSE: {rmse_naive:.4f}")
print(f"Your model RMSE:            {rmse_model:.4f}")
print(f"Skill vs naive: {1 - rmse_model/rmse_naive:.3%}")
```
